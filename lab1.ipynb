```sql
import tensorflow as tf
import numpy as np

# TensorFlow Constant
x = tf.constant([[1., 2., 3.],
                 [4., 5., 6.]])

print(x)
print(x.shape)
print(x.dtype)


# TensorFlow Constants
a = tf.constant(15)
b = tf.constant(20)

print(a+b)


# NumPy Random Array
x = np.random.rand(100).astype(np.float32)
print(x)


# Transformation
y = x * 0.2 + 0.2


# TensorFlow Variables
W = tf.Variable(tf.random.normal([1]))

b = tf.Variable(tf.zeros([1]))


# Linear Regression
import tensorflow as tf

# Variables
W = tf.Variable(0.0)
b = tf.Variable(0.0)

# Loss function
def mse_loss():
    y_pred = W * x + b
    return tf.reduce_mean(tf.square(y_pred - y))

# Optimizer
optimizer = tf.keras.optimizers.Adam(learning_rate=0.01)

# Training loop
for step in range(5000):
    with tf.GradientTape() as tape:
        loss = mse_loss()

    gradients = tape.gradient(loss, [W, b])
    optimizer.apply_gradients(zip(gradients, [W, b]))

    if step % 500 == 0:
        print(f"Step {step}: Loss={loss.numpy():.4f}, W={W.numpy():.4f}, b={b.numpy():.4f}")


# Diabetes Dataset
from numpy import loadtxt
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

dataset = loadtxt('diabetes.csv', delimiter=',', skiprows=1)
dataset


# Pandas
import pandas as pd

df = pd.read_csv('diabetes.csv')
print(df.head())
print(df.columns)


# Split Input and Output
x = dataset[:,0:8]
print(type(x))
print(x.shape)

y = dataset[:,8]
print(y)


# Neural Network Model
model = Sequential()

# The model expects row of data with 8 variables
# 12 = nodes
model.add(Dense(12, input_shape=(8,), activation='relu'))

# Hidden Layer
# 8 = nodes
model.add(Dense(8, activation='relu'))

# Output layer
model.add(Dense(1, activation='sigmoid'))


# Step 3 - Compile the Keras model
# loss (error)
# optimizer (adam)
# metrics = accuracy
model.compile(
    loss='binary_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)


# Step 4 - number of iterations / passes
# 2 - Batch = sample of dataset
model.fit(x, y, epochs=150, batch_size=10)


# Step 5 - evaluate the model
model.evaluate(x, y)
```
